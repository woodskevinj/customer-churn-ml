```python
# ==========================================
# 🧱 Data Preprocessing & Feature Engineering
# ==========================================

# Author: Kevin Woods
# Project: Customer Churn Prediction (Telco)
# ==========================================

import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, StandardScaler
```

```python
# Load Cleaned Data
df = pd.read_csv('../data/raw/WA_Fn-UseC_-Telco-Customer-Churn.csv')
```

```python
# Convert TotalCharges to Numeric (safety check)
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'], errors='coerce')
df['TotalCharges'] = df['TotalCharges'].fillna(df['TotalCharges'].median())
```

```python
# Encode Target Variable
df['Churn'] = df['Churn'].map({'Yes': 1, 'No': 0})
```

```python
# Identify Categorical and Numerical Columns
cat_cols = df.select_dtypes(include=['object']).columns.tolist()
num_cols = df.select_dtypes(include=['int64', 'float64']).columns.tolist()
# remove target if present
num_cols.remove('Churn')

print("Categorical Columns:", cat_cols)
print("Numerical Columns:", num_cols)
```

```python
# One-Hot Encode Categorical Features
df_encoded = pd.get_dummies(df, columns=cat_cols, drop_first=True)

print("Shape after encoding:", df_encoded.shape)
```

```python
# Train-Test Split
X = df_encoded.drop('Churn', axis=1)
y = df_encoded['Churn']

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42
)

print("Training shape:", X_train.shape)
print("Testing shape:", X_test.shape)
```

```python
# Feature Scaling
scaler = StandardScaler()
X_train[num_cols] = scaler.fit_transform(X_train[num_cols])
X_test[num_cols] = scaler.transform(X_test[num_cols])

print("\n✅ Data Preprocessing Complete.")
print(f"Training Samples: {X_train.shape[0]} | Testing Samples: {X_test.shape[0]}")
```

```python
# Save Processed Data
X_train.to_csv('../data/processed/X_train_processed.csv', index=False)
X_test.to_csv('../data/processed/X_test_processed.csv', index=False)
y_train.to_csv('../data/processed/y_train.csv', index=False)
y_test.to_csv('../data/processed/y_test.csv', index=False)
```
