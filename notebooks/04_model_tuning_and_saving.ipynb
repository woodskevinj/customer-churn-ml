```python
# ==========================================
# 🎯 Hyperparameter Tuning & Model Saving
# ==========================================
# Author: Kevin Woods
# Project: Customer Churn Prediction (Telco)
# ==========================================

# --- Imports ---
import pandas as pd
import numpy as np
from sklearn.model_selection import GridSearchCV
from xgboost import XGBClassifier
from sklearn.metrics import classification_report, roc_auc_score
import joblib
```

```python
# ==========================================
# 1️⃣ Load Processed Data
# ==========================================
X_train = pd.read_csv('../data/processed/X_train_processed.csv')
X_test = pd.read_csv('../data/processed/X_test_processed.csv')
y_train = pd.read_csv('../data/processed/y_train.csv').values.ravel()
y_test = pd.read_csv('../data/processed/y_test.csv').values.ravel()

print(f"✅ Data Loaded: {X_train.shape[0]} training samples, {X_test.shape[0]} testing samples")
```

```python
# ==========================================
# 2️⃣ Define Base XGBoost Model
# ==========================================
xgb_base = XGBClassifier(
    objective='binary:logistic',
    use_label_encoder=False,
    eval_metric='logloss',
    random_state=42
)
```

```python
# ==========================================
# 3️⃣ Set Hyperparameter Grid
# ==========================================
param_grid = {
    'n_estimators': [100, 200, 300],
    'max_depth': [3, 4, 5],
    'learning_rate': [0.05, 0.1, 0.2],
    'subsample': [0.8, 1.0],
    'colsample_bytree': [0.8, 1.0]
}
```

```python
# ==========================================
# 4️⃣ Perform Grid Search with Cross-Validation
# ==========================================
print("🚀 Starting hyperparameter tuning (this may take several minutes)...")

grid_search = GridSearchCV(
    estimator=xgb_base,
    param_grid=param_grid,
    scoring='roc_auc',
    cv=3,
    verbose=2,
    n_jobs=-1
)

grid_search.fit(X_train, y_train)

print("\n✅ Grid Search Complete!")
print(f"Best Parameters: {grid_search.best_params_}")
print(f"Best ROC-AUC (CV): {grid_search.best_score_:.4f}")
```

```python
# ==========================================
# 5️⃣ Evaluate Best Model on Test Set
# ==========================================
best_model = grid_search.best_estimator_

y_pred = best_model.predict(X_test)
y_prob = best_model.predict_proba(X_test)[:, 1]

print("\n📊 Final Model Evaluation on Test Data:")
print(classification_report(y_test, y_pred))
print(f"Test ROC-AUC: {roc_auc_score(y_test, y_prob):.4f}")
```

```python
# ==========================================
# 6️⃣ Save the Final Model
# ==========================================
import os
os.makedirs('../models', exist_ok=True)
model_path = '../models/xgb_churn_model.pkl'

joblib.dump(best_model, model_path)
print(f"\n💾 Final model saved successfully: {model_path}")
```
